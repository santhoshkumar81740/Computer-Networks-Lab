```python
import random
import time

def stop_and_wait(frames):
    for frame in frames:
        while True:
            print("\nSending Frame:", frame)
            time.sleep(1)

            # Simulate frame loss
            if random.choice([True, False]):
                print("Frame", frame, "received successfully.")

                # Send acknowledgement
                print("ACK", frame, "received.")
                break
            else:
                print("Frame", frame, "lost!")
                print("Timeout... Retransmitting Frame", frame)

                time.sleep(1)

# Main program
n = int(input("Enter number of frames: "))

frames = list(range(1, n + 1))

print("\n--- Stop and Wait Protocol ---")
stop_and_wait(frames)

print("\nAll frames transmitted successfully!")
```
