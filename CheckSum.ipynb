```python
# Function to add two binary strings using 1's complement addition
def binary_addition(a, b):
    max_len = max(len(a), len(b))
    a = a.zfill(max_len)
    b = b.zfill(max_len)

    result = ''
    carry = 0

    # Perform binary addition from right to left
    for i in range(max_len - 1, -1, -1):
        total = carry + int(a[i]) + int(b[i])
        result = str(total % 2) + result
        carry = total // 2

    # Add carry if present
    if carry:
        result = '1' + result

    # Perform end-around carry (1's complement addition)
    if len(result) > max_len:
        result = binary_addition(result[1:], result[0])

    return result[-max_len:]


# Function to find 1's complement
def ones_complement(binary_str):
    return ''.join('1' if bit == '0' else '0' for bit in binary_str)


# Function to calculate checksum
def calculate_checksum(data_blocks):
    checksum = data_blocks[0]

    for block in data_blocks[1:]:
        checksum = binary_addition(checksum, block)

    checksum = ones_complement(checksum)
    return checksum


# Function to verify checksum at receiver
def verify_checksum(data_blocks, received_checksum):
    total = data_blocks[0]

    for block in data_blocks[1:]:
        total = binary_addition(total, block)

    total = binary_addition(total, received_checksum)

    # If all bits are 1, data is error-free
    return all(bit == '1' for bit in total)


# ---------------- Main Program ----------------

if __name__ == '__main__':

    # Example sender data (8-bit blocks)
    sender_data = [
        '10101010',
        '11001100'
    ]

    print('Sender Data Blocks:')
    for block in sender_data:
        print(block)

    # Calculate checksum
    checksum = calculate_checksum(sender_data)

    print('\nCalculated Checksum:', checksum)

    # Simulate transmission
    transmitted_data = sender_data.copy()
    transmitted_data.append(checksum)

    print('\nData Sent:')
    for block in transmitted_data:
        print(block)

    # Receiver verification
    print('\nReceiver Side Verification')

    if verify_checksum(sender_data, checksum):
        print('Data received correctly. No error detected.')
    else:
        print('Error detected in received data.')
```
