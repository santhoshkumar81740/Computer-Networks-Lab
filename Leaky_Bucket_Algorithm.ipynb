```python
def leaky_bucket():
    bucket_size = int(input("Enter bucket size: "))
    output_rate = int(input("Enter output rate: "))
    n = int(input("Enter number of packets: "))

    bucket = 0

    for i in range(1, n + 1):
        packet_size = int(input(f"Enter size of packet {i}: "))

        # Check for overflow
        if bucket + packet_size > bucket_size:
            print("Packet", i, "dropped due to bucket overflow")
        else:
            bucket += packet_size
            print("Packet", i, "added to bucket")

        # Send packets at fixed rate
        if bucket >= output_rate:
            bucket -= output_rate
            print("Output:", output_rate, "units")
        else:
            print("Output:", bucket, "units")
            bucket = 0

        print("Remaining in bucket:", bucket)
        print()

    # Empty remaining packets
    while bucket > 0:
        if bucket >= output_rate:
            bucket -= output_rate
            print("Output:", output_rate, "units")
        else:
            print("Output:", bucket, "units")
            bucket = 0

        print("Remaining in bucket:", bucket)


# Main program
print("----- Leaky Bucket Algorithm -----")
leaky_bucket()
```
